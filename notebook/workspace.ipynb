```python
# Check what ID that are not exist in sample_sub_reg.csv that in test.csv

import pandas as pd

# Load the test data
test_data = pd.read_csv("data/test.csv")

print("Total records: ", len(test_data))

# Load the sample submission data
sample_submission = pd.read_csv("data/sample_sub_reg.csv")

print("Total records: ", len(sample_submission))

# Find IDs that exist in test_data but not in sample_submission

missing_ids = test_data[~test_data['id'].isin(sample_submission['Id'])]

print(missing_ids)

# Save the missing IDs to a CSV file
#missing_ids.to_csv("missing_ids.csv", index=False)
```
